```python
import wandb
import matplotlib.pyplot as plt

api = wandb.Api()
# run = api.run("tbal/AdEMAMIX_adaptive/53bqfvwf")
run = api.run("tbal/AdEMAMIX_adaptive/dcupgs5b")

history1 = run.history(keys=["iter", "two_plane_momo/network/<g_t, m_t^(1)>"])
history2 = run.history(keys=["iter", "two_plane_momo/network/<g_t, m_t^(2)>"])
history3 = run.history(keys=["iter", "two_plane_momo/network/<m_t^(1), m_t^(2)>"])

plt.plot(history1["iter"], history1["two_plane_momo/network/<g_t, m_t^(1)>"], label="<g_t, m_t^(1)>", alpha=0.5)
plt.plot(history2["iter"], history2["two_plane_momo/network/<g_t, m_t^(2)>"], label="<g_t, m_t^(2)>", alpha=0.5)
plt.plot(history3["iter"], history3["two_plane_momo/network/<m_t^(1), m_t^(2)>"], label="<m_t^(1), m_t^(2)>", alpha=0.5)
plt.xlabel("iter")
plt.ylabel("<g_t, m_t^(1)> and <g_t, m_t^(2)>")
plt.legend()
plt.show()
```

```python
history1 = run.history(keys=["iter", "two_plane_momo/network/cos_sim(g_t, m_t^(1))"])
history2 = run.history(keys=["iter", "two_plane_momo/network/cos_sim(g_t, m_t^(2))"])
history3 = run.history(keys=["iter", "two_plane_momo/network/cos_sim(m_t^(1), m_t^(2))"])

plt.plot(history1["iter"], history1["two_plane_momo/network/cos_sim(g_t, m_t^(1))"], label="cos_sim(g_t, m_t^(1))", alpha=0.5)
plt.plot(history2["iter"], history2["two_plane_momo/network/cos_sim(g_t, m_t^(2))"], label="cos_sim(g_t, m_t^(2))", alpha=0.5)
plt.plot(history3["iter"], history3["two_plane_momo/network/cos_sim(m_t^(1), m_t^(2))"], label="cos_sim(m_t^(1), m_t^(2))", alpha=0.5)

plt.xlabel("iter")
plt.ylabel("cos_sim(g_t, m_t^(1)) vs cos_sim(g_t, m_t^(2))")
plt.legend()
plt.show()
```

```python
history1 = run.history(keys=["iter", "two_plane_momo/network/||g_t||_{2}^{2}"])
history2 = run.history(keys=["iter", "two_plane_momo/network/||m_t^{(1)}||_{2}^{2}"])
history3 = run.history(keys=["iter", "two_plane_momo/network/||m_t^{(2)}||_{2}^{2}"])

plt.plot(history1["iter"], history1["two_plane_momo/network/||g_t||_{2}^{2}"], label="squared norm g_t", alpha=0.5)
plt.plot(history2["iter"], history2["two_plane_momo/network/||m_t^{(1)}||_{2}^{2}"], label="squared norm m_t^{(1)}", alpha=0.5)
plt.plot(history3["iter"], history3["two_plane_momo/network/||m_t^{(2)}||_{2}^{2}"], label="squared norm m_t^{(2)}", alpha=0.5)
plt.xlabel("iter")
plt.ylabel("squared norms of g_t, m_t^(1), and m_t^(2)")
plt.legend()
plt.show()
```

```python
history1 = run.history(keys=["iter", "two_plane_momo/network/||g_t||_{2}^{2}"]).dropna()
history2 = run.history(keys=["iter", "two_plane_momo/network/||m_t^{(1)}||_{2}^{2}"]).dropna()
history3 = run.history(keys=["iter", "two_plane_momo/network/||m_t^{(2)}||_{2}^{2}"]).dropna()

plt.plot(history1["iter"], np.sqrt(history1["two_plane_momo/network/||g_t||_{2}^{2}"]), label="norm g_t", alpha=0.5)
plt.plot(history2["iter"], np.sqrt(history2["two_plane_momo/network/||m_t^{(1)}||_{2}^{2}"]), label="norm m_t^{(1)}", alpha=0.5)
plt.plot(history3["iter"], np.sqrt(history3["two_plane_momo/network/||m_t^{(2)}||_{2}^{2}"]), label="norm m_t^{(2)}", alpha=0.5)
plt.xlabel("iter")
plt.ylabel("norms of g_t, m_t^(1), and m_t^(2)")
plt.legend()
plt.show()
```

```python
history1 = run.history(keys=["iter", "two_plane_momo/network/numer_term_A_num_fac_times_b_gap"]).dropna()
history2 = run.history(keys=["iter", "two_plane_momo/network/numer_term_B_mu_times_m1_minus_m2_dot_wt"]).dropna()
history3 = run.history(keys=["iter", "two_plane_momo/network/numer_term_C_m1_minus_m2_dot_Pinv_m2"]).dropna()
history4 = run.history(keys=["iter", "two_plane_momo/network/corrected_denom"]).dropna()

plt.plot(history1["iter"], history1["two_plane_momo/network/numer_term_A_num_fac_times_b_gap"], label="numer_termA", alpha=0.5)
plt.plot(history2["iter"], history2["two_plane_momo/network/numer_term_B_mu_times_m1_minus_m2_dot_wt"], label="numer_termB", alpha=0.5)
plt.plot(history3["iter"], history3["two_plane_momo/network/numer_term_C_m1_minus_m2_dot_Pinv_m2"], label="numer_termC", alpha=0.5)
plt.plot(history4["iter"], history4["two_plane_momo/network/corrected_denom"], label="corrected_denom", alpha=0.5)

plt.xlabel("iter")
plt.ylabel("numerator terms to compute alpha and corrected denominator")
plt.legend()
plt.show()
```

```python
history1 = run.history(keys=["iter", "two_plane_momo/network/barf1"]).dropna()
history2 = run.history(keys=["iter", "two_plane_momo/network/gamma1"]).dropna()
history3 = run.history(keys=["iter", "two_plane_momo/network/<m_t^(1), w_t>"]).dropna()
history4 = run.history(keys=["iter", "two_plane_momo/network/b1"]).dropna()

history5 = run.history(keys=["iter", "two_plane_momo/network/barf2"]).dropna()
history6 = run.history(keys=["iter", "two_plane_momo/network/gamma2"]).dropna()
history7 = run.history(keys=["iter", "two_plane_momo/network/<m_t^(2), w_t>"]).dropna()
history8 = run.history(keys=["iter", "two_plane_momo/network/b2"]).dropna()


plt.plot(history1["iter"], history1["two_plane_momo/network/barf1"], label="barf1", alpha=0.5)
plt.plot(history2["iter"], history2["two_plane_momo/network/gamma1"], label="gamma1", alpha=0.5)
plt.plot(history3["iter"], history3["two_plane_momo/network/<m_t^(1), w_t>"], label="<m_t^(1), w_t>", alpha=0.5)
plt.plot(history4["iter"], history4["two_plane_momo/network/b1"], label="b1_t = barf1_t - gamma1_t + <m_t^(1), w_t>", alpha=0.5)

plt.plot(history5["iter"], history5["two_plane_momo/network/barf2"], label="barf2", alpha=0.5)
plt.plot(history6["iter"], history6["two_plane_momo/network/gamma2"], label="gamma2", alpha=0.5)
plt.plot(history7["iter"], history7["two_plane_momo/network/<m_t^(2), w_t>"], label="<m_t^(2), w_t>", alpha=0.5)
plt.plot(history8["iter"], history8["two_plane_momo/network/b2"], label="b2_t = barf2_t - gamma2_t + <m_t^(2), w_t>", alpha=0.5)

plt.xlabel("iter")
plt.ylabel("value")
plt.legend()
plt.show()
```
